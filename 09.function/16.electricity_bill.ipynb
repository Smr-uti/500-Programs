#### Write a Python program using a function to calculate the total electricity bill.

```python
def calculate_electricity_bill(units):

    if units <= 0:
        return "Invalid Units"
    elif units <= 100:
       rate = 5
    elif units <= 200:
        rate = 7 
    else:
        rate = 10

    bill = units * rate
    total_bill = bill + 100

    return total_bill

units=int(input("Enter units: "))

result = calculate_electricity_bill(units)

print(result)
```
