## Get First 2 and Last 2 Characters
#### Problem Statement:

##### Write a Python function to create a new string from the first 2 and last 2 characters of a given string. If the string length is less than 2, return an empty string.

```python
def extract_edges(input_str):
    if (len(input_str) < 2 ):
        return ""
    return input_str[0:2]+input_str[9:11]


input_string=input("Enter the string:")
result=extract_edges(input_string)
print("new string is:",result)
```
