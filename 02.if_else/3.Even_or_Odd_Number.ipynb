#### Write a Python program that asks the user for a number and determines whether it is an Even number or an Odd number.

```python
print("Program Starts")

num=int(input("Enter any Number:"))
print(num)

if num % 2 ==0:
    print("Number is Even")
else:
    print("Number is Odd")

print("Program Ends")
```
