### Write a Python function that replaces all occurrences of the first character with '$', except the first character itself.

```python
def replace_first_char(text):
    first_char=text[0]
    modified_text=text.replace(first_char,"$")
    new_modified_list=first_char+modified_text[1:]
    return new_modified_list

input_str="refresh"
result=replace_first_char(input_str)
print("New String is:",result)



```
