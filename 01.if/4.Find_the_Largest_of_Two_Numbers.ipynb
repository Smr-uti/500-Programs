#### Write a Python program to take two different numbers as input from the user and find the maximum (largest) number between them.

```python
print("Program Starts")

num1=int(input("Enter any number"))

num2=int(input("Enter any number"))


if num1>num2:
    print(f"{num1} is the largest number.")
else:
    print(f"{num2} is the largest number.")
    

print("Program Ends")
```
