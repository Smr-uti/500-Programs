#### Write a Python program to accept a person's age as input. Check if the person is eligible to vote or not (the minimum age requirement for voting is 18 years).

```python
print("Program Starts")

age=int(input("Enter your age:"))
print(age)

if age>=18:
    print("Your are eligible for voting:")
else:
    print("You are not eligible for voting")

print("Program Ends")
```
